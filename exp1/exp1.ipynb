Numpy

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
# 1D array
arr1 = np.array([1, 2, 3, 4, 5])
print("Original array:", arr1)
print("Size of the array:", arr1.size)
print("Shape of the array:", arr1.shape)
print("Dimensions of the array:", arr1.ndim)
```

```python
#2D array
arr2 = np.array([[1, 2, 3], [4, 5, 6]])
print("Original 2D array:\n", arr2)
print("Size of the array:", arr2.size)
print("Shape of the array:", arr2.shape)
print("Dimensions of the array:", arr2.ndim)
```

```python
plt.plot(arr1, marker='o', linestyle='-', color='b')
plt.title("1D Array Plot")
plt.xlabel("Index")
plt.ylabel("Value")
```

```python
plt.plot(arr2, color='r')
plt.title("2D Array Plot")
plt.xlabel("Index")
plt.ylabel("Value")
```

```python
# Random array
random_array = np.random.rand(3, 4)
print("Random array:\n", random_array)
print("Size of the random array:", random_array.size)
print("Shape of the random array:", random_array.shape)
print("Dimensions of the random array:", random_array.ndim)
```

```python
r1 = np.random.rand(3)
print("Random array r1:\n", r1)
r2 = np.random.random((3, 4))
print("Random array r2:\n", r2)
r3 = np.random.randint(1, 10,size=(3, 4))
print("Random array r3:\n", r3)
```

```python
# Array with arange function
arange_array = np.arange(0, 10, 2)
print("Array with arange function:", arange_array)
print("Size of the arange array:", arange_array.size)
```

```python
arr3 = np.linspace(0, 10, 5)
print("Array :", arr3)
```

```python
plt.plot(r3)
plt.plot(arr3)
```
