```python
import matplotlib.pyplot as plt
import seaborn as sns

plt.figure(figsize=(12,3))   # Wide and short figure

sns.boxplot(
    x=df["Income"],
    color="skyblue",
    width=0.4
)

plt.title("Horizontal Boxplot of Income with outliers", fontsize=14)
plt.xlabel("Income")
plt.grid(axis='x', linestyle='--', alpha=0.6)

plt.savefig("boxplot_income_outliers.png")
plt.show()
```

```python
Q1 = df["Income"].quantile(0.25)
Q3 = df["Income"].quantile(0.75)

print("Q1 =", Q1)
print("Q3 =", Q3)
```

```python
IQR = Q3 - Q1

print("IQR =", IQR)
```

```python
lower_limit = Q1 - 1.5 * IQR
upper_limit = Q3 + 1.5 * IQR

print("Lower Limit =", lower_limit)
print("Upper Limit =", upper_limit)
```

```python
outliers = df[(df["Income"] < lower_limit) | (df["Income"] > upper_limit)]

print("Number of outliers:", len(outliers))
print("Outliers:")
print(outliers)
```

```python
df = df[(df["Income"] >= lower_limit) & (df["Income"] <= upper_limit)]
```

```python
df.shape
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

plt.figure(figsize=(10,4))

sns.boxplot(x=df["Income"])

plt.title("Income After Removing Outliers")
plt.savefig("boxplot_income_no_outliers.png")
plt.show()
```

```python
remaining = df[(df["Income"] < lower_limit) | (df["Income"] > upper_limit)]

print("Remaining Outliers:", len(remaining))
```

```python
plt.figure(figsize=(8,5))
sns.histplot(df["Income"], bins=30, kde=True)

plt.title("Distribution of Income")
plt.xlabel("Income")
plt.ylabel("Frequency")
plt.savefig("histogram_income.png")
plt.show()
```

```python
plt.figure(figsize=(8,5))

sns.countplot(x="Marital_Status", data=df)

plt.xticks(rotation=45)

plt.title("Marital Status Distribution")

plt.savefig("countplot_marital_status.png")
plt.show()
```

```python
education = df["Education"].value_counts()

plt.figure(figsize=(7,7))

plt.pie(
    education,
    labels=education.index,
    autopct="%1.1f%%",
    startangle=90
)

plt.title("Education Percentage")
plt.savefig("piechart_education.png")
plt.show()
```

```python
avg = df.groupby("Kidhome")["MntWines"].mean().reset_index()

plt.figure(figsize=(6,4))

sns.lineplot(
    x="Kidhome",
    y="MntWines",
    data=avg,
    marker="o"
)

plt.title("Wine Spending by Number of Kids")
plt.savefig("lineplot_wine_spending.png")
plt.show()
```
