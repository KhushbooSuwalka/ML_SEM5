```python
import pandas as pd
import math
from sklearn.model_selection import train_test_split
from sklearn.tree import DecisionTreeClassifier
from sklearn.metrics import accuracy_score
```

```python
# Dataset
data = {
    "Outlook": ["Sunny", "Sunny", "Overcast", "Rain", "Rain",
                "Rain", "Overcast", "Sunny", "Sunny", "Rain",
                "Sunny", "Overcast", "Overcast", "Rain"],

    "Temperature": ["Hot", "Hot", "Hot", "Mild", "Cool",
                    "Cool", "Cool", "Mild", "Cool", "Mild",
                    "Mild", "Mild", "Hot", "Mild"],

    "Humidity": ["High", "High", "High", "High", "Normal",
                "Normal", "Normal", "High", "Normal", "Normal",
                "Normal", "High", "Normal", "High"],

    "Wind": ["Weak", "Strong", "Weak", "Weak", "Weak",
             "Strong", "Strong", "Weak", "Weak", "Weak",
             "Strong", "Strong", "Weak", "Strong"],

    "Play": ["No", "No", "Yes", "Yes", "Yes", "No",
             "Yes", "No", "Yes", "Yes", "Yes", "Yes",
             "Yes", "No"]
}

df = pd.DataFrame(data)
```

```python
# 1. Entropy function
def entropy(y):
    counts = y.value_counts()
    total = len(y)
    ent = 0

    for count in counts:
        p = count / total
        ent -= p * math.log2(p)

    return ent
```

```python
# 2. Information Gain function
def information_gain(df, feature, target):
    total_entropy = entropy(df[target])
    weighted_entropy = 0

    for value in df[feature].unique():
        subset = df[df[feature] == value]
        weighted_entropy += (len(subset) / len(df)) * entropy(subset[target])

    return total_entropy - weighted_entropy
```

```python
# 3. Calculate Information Gain for all features
features = ["Outlook", "Temperature", "Humidity", "Wind"]

for feature in features:
    print(feature, "Information Gain:", information_gain(df, feature, "Play"))
```

```python
# 4. Separate X and y
X = df[features]
y = df["Play"]
```

```python
# 5. Encode categorical data
X = pd.get_dummies(X)
```

```python
# 6. Train-test split
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.3, random_state=42
)

# 7. Decision Tree model using Entropy
model = DecisionTreeClassifier(
    criterion="entropy",
    random_state=42
)

# 8. Train model
model.fit(X_train, y_train)

# 9. Prediction
y_pred = model.predict(X_test)
```

```python
# 10. Accuracy
accuracy = accuracy_score(y_test, y_pred)

print("\nActual Values:", list(y_test))
print("Predicted Values:", list(y_pred))
print("Accuracy:", accuracy * 100, "%")
```
