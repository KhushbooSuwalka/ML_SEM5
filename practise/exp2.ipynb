```python
import pandas as pd
import numpy as np
from sklearn.preprocessing import StandardScaler,MinMaxScaler,OneHotEncoder
```

```python
# Create a dataset
df = pd.DataFrame({
    "Name": ["A", "B", "C", "D"],
    "Age": [20, 21, 19, np.nan],
    "Marks": [80, np.nan, 70, 90],
    "City": ["Jaipur", "Delhi", "Jaipur", "Mumbai"]
})

print("Original Data:")
print(df)
```

```python
# 1. Handling missing values
df["Age"] = df["Age"].fillna(df["Age"].mean())
df["Marks"] = df["Marks"].fillna(df["Marks"].mean())
```

```python
# 2. Filtering data
filtered_df = df[df["Marks"] > 75]
print("Filtered Data:")
print(filtered_df)
```

```python
# 3. Reshaping
reshaped_df = df.pivot_table(index="City", values="Marks", aggfunc="mean")
print("Reshaped Data:")
print(reshaped_df)
```

```python
# 4.Merging data
extra = pd.DataFrame({
    "Name": ["E", "F"],
    "Age": [22, 23]
})
merged_df = pd.merge(df, extra, on="Name", how="outer")
print("Merged Data:")   
print(merged_df)
```

```python
# 5.Scaling numeric data
scaler = StandardScaler()
scaled_data = scaler.fit_transform(df[["Marks"]])
print("Scaled Data:")
print(scaled_data)
```

```python
scaler = MinMaxScaler()
scaled_data = scaler.fit_transform(df[["Marks"]])
print("Min-Max Scaled Data:")
print(scaled_data)
```

```python
# 6.Encoding categorical data
encoder = pd.get_dummies(df["City"], dtype=int)
print("Encoded Data:")
print(encoder)
```
